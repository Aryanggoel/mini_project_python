This beginner-level project calculates an approximate Air Quality Index (AQI) from pollutant concentrations using Python, NumPy and Pandas.


## 1. Import Libraries
We use **NumPy** for numerical calculations and **Pandas** for storing and handling pollutant data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Enter Pollutant Values
Enter sample concentrations for PM2.5, PM10, NO2, SO2 and CO. These are example values for a college mini project.

```python
pollutants = {
    'PM2.5': 65,
    'PM10': 120,
    'NO2': 45,
    'SO2': 30,
    'CO': 1.2
}

df = pd.DataFrame([pollutants])
df
```

## 3. AQI Calculation Function
For simplicity, this project uses a **beginner-friendly linear scale** for each pollutant. The project is intended for learning and is not an official regulatory AQI calculator.

```python
def calculate_sub_index(value, good, poor):
    """Convert a pollutant value to a simple 0-500 learning scale."""
    index = (value - good) / (poor - good) * 100
    return np.clip(index, 0, 500)

df['PM2.5_AQI'] = calculate_sub_index(df['PM2.5'], 0, 250)
df['PM10_AQI'] = calculate_sub_index(df['PM10'], 0, 430)
df['NO2_AQI'] = calculate_sub_index(df['NO2'], 0, 400)
df['SO2_AQI'] = calculate_sub_index(df['SO2'], 0, 200)
df['CO_AQI'] = calculate_sub_index(df['CO'], 0, 10)

df
```

## 4. Find Overall AQI
The overall AQI is taken as the **maximum pollutant sub-index** in this simplified project.

```python
aqi_columns = ['PM2.5_AQI', 'PM10_AQI', 'NO2_AQI', 'SO2_AQI', 'CO_AQI']

overall_aqi = df[aqi_columns].max(axis=1).iloc[0]
overall_aqi = round(float(overall_aqi), 2)

print('Overall AQI:', overall_aqi)
```

## 5. AQI Category
The following categories are used for easy project interpretation.

```python
def aqi_category(aqi):
    if aqi <= 50:
        return 'Good'
    elif aqi <= 100:
        return 'Satisfactory'
    elif aqi <= 200:
        return 'Moderate'
    elif aqi <= 300:
        return 'Poor'
    elif aqi <= 400:
        return 'Very Poor'
    else:
        return 'Severe'

category = aqi_category(overall_aqi)
print('AQI Category:', category)
```

## 6. Display Final Result


```python
print('===== AQI CALCULATOR =====')
print('Student: Aryan Goel')
print('Overall AQI:', overall_aqi)
print('Category:', category)
print('\nPollutant-wise sub-index:')
print(df[aqi_columns].T.rename(columns={0: 'AQI'}))
```

## 7. Enter Your Own Values
Run the cell below and enter your own pollutant values when Colab asks for them. The program will calculate the AQI and category automatically.


```python
print('===== ENTER POLLUTANT VALUES =====')
print('Enter numeric values for each pollutant.')

try:
    pm25 = float(input('Enter PM2.5 value: '))
    pm10 = float(input('Enter PM10 value: '))
    no2 = float(input('Enter NO2 value: '))
    so2 = float(input('Enter SO2 value: '))
    co = float(input('Enter CO value: '))

    user_values = {
        'PM2.5': pm25,
        'PM10': pm10,
        'NO2': no2,
        'SO2': so2,
        'CO': co
    }

    user_df = pd.DataFrame([user_values])

    poor_limits = {
        'PM2.5': 250,
        'PM10': 430,
        'NO2': 400,
        'SO2': 200,
        'CO': 10
    }

    for pollutant, poor_limit in poor_limits.items():
        user_df[pollutant + '_AQI'] = calculate_sub_index(
            user_df[pollutant], 0, poor_limit
        )

    user_aqi = round(float(user_df[aqi_columns].max(axis=1).iloc[0]), 2)
    user_category = aqi_category(user_aqi)

    print('\n===== USER RESULT =====')
    print('Overall AQI:', user_aqi)
    print('AQI Category:', user_category)
    
    print('\nPollutant-wise AQI:')
    print(user_df[aqi_columns].T.rename(columns={0: 'AQI'}))

except ValueError:
    print('Please enter valid numeric values only.')
```

## 8. AQI Graph
The bar graph below shows the AQI sub-index contributed by each pollutant in the original sample data.


```python
plt.figure(figsize=(9, 5))
plt.bar(df[aqi_columns].columns.str.replace('_AQI', ''), df[aqi_columns].iloc[0])
plt.title('Pollutant-wise AQI Sub-index')
plt.xlabel('Pollutants')
plt.ylabel('AQI Sub-index')
plt.ylim(0, 500)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()
```
